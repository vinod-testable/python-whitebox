```python
from radon import complexity, metrics, raw, visitors
print('complexity:', dir(complexity))
print('metrics:', dir(metrics))
print('raw:', dir(raw))
print('visitors:', dir(visitors))
```

```python
# Raw JSON from radon cc (cyclomatic complexity)
import subprocess, os

rw = os.path.join(os.path.dirname(os.getcwd()), 'redditwarp', 'redditwarp')
r = subprocess.run(['radon', 'cc', '-j', '-a', '-s', rw], capture_output=True, text=True, timeout=120)
print(r.stdout)
```
