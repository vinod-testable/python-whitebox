# SpotBugs — Bytecode Bug Detection for Java

[SpotBugs](https://spotbugs.github.io/) (successor to FindBugs) analyses Java **bytecode** (`.class` / `.jar`) to detect patterns likely to be bugs.

## What SpotBugs produces
- **Bug instances** with: bug type, category, priority, class, method, line.
- **Bug categories**: `CORRECTNESS`, `MT_CORRECTNESS`, `BAD_PRACTICE`, `PERFORMANCE`, `SECURITY`, `STYLE`, `EXPERIMENTAL`.
- Formats: `xml` (with messages), `html`, `text`, `emacs`, `xdocs`, `sarif`.

## Important
SpotBugs requires **compiled bytecode** — point it at a `target/` folder, `.jar` file, or `out/` directory **not** at `.java` sources.

## CLI
```
spotbugs -textui -effort:max [-xml:withMessages] [-html] -output <file> <class-dir-or-jar>
```

## 1. Configuration

````python
from __future__ import annotations
import os, subprocess, sys
from pathlib import Path

_base = Path(r"F:\java_metrics")

# Point to compiled bytecode: a .jar, or a directory containing .class files
ANALYZE_TARGET = Path(
    os.environ.get("SPOTBUGS_TARGET", _base / "sample-java-app" / "out" / "app.jar")
).resolve()

SPOTBUGS_HOME  = Path(
    os.environ.get("SPOTBUGS_HOME", _base / "tools" / "spotbugs-4.9.8")
).resolve()
OUTPUT_DIR     = _base / "spotbugs_out"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

SB_EXE = SPOTBUGS_HOME / "bin" / ("spotbugs.bat" if sys.platform == "win32" else "spotbugs")
EFFORT = "max"   # min | default | max

assert SB_EXE.is_file(),        f"spotbugs not found: {SB_EXE}"
assert ANALYZE_TARGET.exists(), f"Bytecode target not found: {ANALYZE_TARGET}"
print(f"SB_EXE         : {SB_EXE}")
print(f"ANALYZE_TARGET : {ANALYZE_TARGET}")
print(f"OUTPUT_DIR     : {OUTPUT_DIR}")
````

## 2. Show SpotBugs version

````python
proc = subprocess.run(
    [str(SB_EXE), "-version"],
    capture_output=True, text=True
)
print(proc.stdout or proc.stderr)
````

## 3. Run SpotBugs — XML with messages (primary format)

````python
xml_report = OUTPUT_DIR / "spotbugs-report.xml"
cmd = [
    str(SB_EXE),
    "-textui",
    f"-effort:{EFFORT}",
    "-xml:withMessages",
    "-output", str(xml_report),
    str(ANALYZE_TARGET),
]
print("Command:", " ".join(cmd))
proc_xml = subprocess.run(cmd, capture_output=True, text=True)
print("\n=== STDOUT ===")
print(proc_xml.stdout or "(empty)")
print("\n=== STDERR ===")
print(proc_xml.stderr or "(empty)")
# 0 = no bugs, 1 = bugs found, 2 = error
print("\nExit code:", proc_xml.returncode,
      "(0 = no bugs, 1 = bugs found, 2 = error)")
````

## 4. Run SpotBugs — HTML report

````python
html_report = OUTPUT_DIR / "spotbugs-report.html"
cmd_html = [
    str(SB_EXE),
    "-textui",
    f"-effort:{EFFORT}",
    "-html",
    "-output", str(html_report),
    str(ANALYZE_TARGET),
]
print("Command:", " ".join(cmd_html))
proc_html = subprocess.run(cmd_html, capture_output=True, text=True)
print("STDOUT:", proc_html.stdout or "(empty)")
print("STDERR:", proc_html.stderr or "(empty)")
print("Exit code:", proc_html.returncode)
````

## 5. Run SpotBugs — plain text report

````python
text_report = OUTPUT_DIR / "spotbugs-report.txt"
cmd_txt = [
    str(SB_EXE),
    "-textui",
    f"-effort:{EFFORT}",
    "-output", str(text_report),
    str(ANALYZE_TARGET),
]
print("Command:", " ".join(cmd_txt))
proc_txt = subprocess.run(cmd_txt, capture_output=True, text=True)
print("STDOUT:", proc_txt.stdout or "(empty)")
print("STDERR:", proc_txt.stderr or "(empty)")
print("Exit code:", proc_txt.returncode)
````

## 6. Raw XML report

````python
if xml_report.is_file():
    raw = xml_report.read_text(encoding="utf-8", errors="replace")
    print(f"=== {xml_report.name} ===")
    print(raw)
else:
    print("XML report not produced.")
````

## 7. Raw text report

````python
if text_report.is_file():
    content = text_report.read_text(encoding="utf-8", errors="replace")
    print(f"=== {text_report.name} ===")
    print(content or "(empty — no bugs found)")
else:
    print("Text report not produced.")
````

## 8. HTML report path

````python
if html_report.is_file():
    print("Open in browser:", html_report.resolve().as_uri())
else:
    print("HTML report not produced.")
````

## 9. All output files

````python
print("Files in output directory:")
for f in sorted(OUTPUT_DIR.iterdir()):
    print(f"  {f.name}  ({f.stat().st_size:,} bytes)")
````
