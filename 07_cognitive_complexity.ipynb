```python
# Raw cognitive complexity scores for all functions in redditwarp — full JSON output
import ast, os, json
from cognitive_complexity.api import get_cognitive_complexity

rw = os.path.join(os.path.dirname(os.getcwd()), 'redditwarp', 'redditwarp')
output = []

for root, dirs, files in os.walk(rw):
    dirs[:] = [d for d in dirs if d != '__pycache__']
    for fname in files:
        if not fname.endswith('.py'):
            continue
        fpath = os.path.join(root, fname)
        try:
            src = open(fpath, encoding='utf-8', errors='ignore').read()
            tree = ast.parse(src)
            for node in ast.walk(tree):
                if isinstance(node, (ast.FunctionDef, ast.AsyncFunctionDef)):
                    score = get_cognitive_complexity(node)
                    output.append({
                        'file': os.path.relpath(fpath, rw),
                        'function': node.name,
                        'lineno': node.lineno,
                        'col_offset': node.col_offset,
                        'is_async': isinstance(node, ast.AsyncFunctionDef),
                        'cognitive_complexity': score
                    })
        except Exception:
            pass

print(json.dumps(output, indent=2))
```
