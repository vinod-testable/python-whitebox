# pip-audit — Python Dependency Vulnerability Scanner
Install: `pip install pip-audit` | CLI: `pip-audit --format json`

```python
import pip_audit
print(dir(pip_audit))
```

```python
import pkgutil, importlib
for finder, name, ispkg in pkgutil.walk_packages(pip_audit.__path__, pip_audit.__name__ + '.', onerror=lambda x: None):
    try:
        mod = importlib.import_module(name)
        print(name, '->', dir(mod))
    except Exception as e:
        print(name, '-> SKIP:', e)
```

```python
# Raw JSON output — audit current environment
import subprocess

r = subprocess.run(
    ['pip-audit', '--format', 'json', '--progress-spinner', 'off'],
    capture_output=True, text=True, encoding='utf-8', errors='replace', timeout=180
)
print(r.stdout)
```

```python
# Raw pip-audit against redditwarp requirements
# Note: redditwarp uses Poetry-format pyproject.toml (not PEP 621), so we use
# the docs/requirements.txt; fall back to --local if it doesn't exist.
import subprocess, os

rw = os.path.join(os.path.dirname(os.getcwd()), 'redditwarp')
req = os.path.join(rw, 'docs', 'requirements.txt')

if os.path.exists(req):
    r = subprocess.run(
        ['pip-audit', '--format', 'json', '--progress-spinner', 'off', '-r', req],
        capture_output=True, text=True, encoding='utf-8', errors='replace', timeout=180
    )
else:
    r = subprocess.run(
        ['pip-audit', '--format', 'json', '--progress-spinner', 'off', '--local'],
        capture_output=True, text=True, encoding='utf-8', errors='replace', timeout=180
    )
if r.stderr:
    print('stderr:', r.stderr[:500])
print(r.stdout)
```
