```python
# Raw lizard XML output (native rich format)
import subprocess, os

rw = os.path.join(os.path.dirname(os.getcwd()), 'redditwarp', 'redditwarp')
r = subprocess.run(['lizard', rw, '--xml'], capture_output=True, text=True, timeout=120)
print(r.stdout)
```

```python
# Raw all FileInformation and FunctionInfo fields as JSON
import lizard, os, json

rw = os.path.join(os.path.dirname(os.getcwd()), 'redditwarp', 'redditwarp')
output = []

for file_info in lizard.analyze([rw]):
    file_record = {
        'filename': file_info.filename,
        'nloc': file_info.nloc,
        'token_count': file_info.token_count,
        'CCN': file_info.CCN,
        'ND': file_info.ND,
        'average_nloc': file_info.average_nloc,
        'average_token_count': file_info.average_token_count,
        'average_cyclomatic_complexity': file_info.average_cyclomatic_complexity,
        'functions': [
            {
                'name': f.name,
                'long_name': f.long_name,
                'filename': f.filename,
                'nloc': f.nloc,
                'cyclomatic_complexity': f.cyclomatic_complexity,
                'token_count': f.token_count,
                'parameters': f.parameters,
                'parameter_count': f.parameter_count,
                'start_line': f.start_line,
                'end_line': f.end_line,
                'length': f.length,
                'max_nesting_depth': f.max_nesting_depth,
                'top_nesting_level': f.top_nesting_level,
            }
            for f in file_info.function_list
        ]
    }
    output.append(file_record)

print(json.dumps(output, indent=2))
```
