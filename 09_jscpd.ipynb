# jscpd — Copy/Paste Detector (Multi-Language)
Install: `npm install -g jscpd` | CLI: `jscpd <path> --reporters json`

```python
import subprocess
r = subprocess.run(['jscpd', '--version'], capture_output=True, text=True, shell=True)
print('jscpd version:', r.stdout.strip())
print()
r2 = subprocess.run(['jscpd', '--help'], capture_output=True, text=True, shell=True)
print(r2.stdout)
```

```python
# Raw jscpd --list output (all supported formats/languages)
import subprocess
r = subprocess.run(['jscpd', '--list'], capture_output=True, text=True, shell=True)
print(r.stdout)
```

```python
# JSON report from jscpd on jscpd_duplication_sample.py
# Windows: jscpd ignores absolute paths with backslashes; use Path.as_posix() for the scan path.
import json
import os
import shutil
import subprocess
import tempfile
from pathlib import Path

target = (Path.cwd() / "jscpd_duplication_sample.py").resolve()
scan_path = target.as_posix()  # e.g. D:/proj/file.py — required on Windows for jscpd to scan
report_dir = tempfile.mkdtemp()
jscpd = shutil.which("jscpd")
if not jscpd:
    raise FileNotFoundError("jscpd not on PATH (npm install -g jscpd)")

r = subprocess.run(
    [
        jscpd,
        scan_path,
        "--reporters",
        "json",
        "--output",
        report_dir,
        "--format",
        "python",
        "--min-lines",
        "5",
        "--min-tokens",
        "50",
    ],
    capture_output=True,
    text=True,
    timeout=120,
    shell=False,
)
print("RETURNCODE:", r.returncode)
print("STDOUT:\n", r.stdout)
if r.stderr.strip():
    print("STDERR:\n", r.stderr)

json_path = os.path.join(report_dir, "jscpd-report.json")
if os.path.exists(json_path):
    with open(json_path, encoding="utf-8") as f:
        raw = json.load(f)
    print(json.dumps(raw, indent=2))
else:
    print("No jscpd-report.json in", report_dir)
    print("Contents:", os.listdir(report_dir))
```
