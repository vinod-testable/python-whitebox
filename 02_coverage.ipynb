```python
# Collect coverage on sample code then emit full raw JSON report
import tempfile, os, sys, importlib, json
from coverage import Coverage

tmp_dir = tempfile.mkdtemp()
mod_path = os.path.join(tmp_dir, 'sample.py')
with open(mod_path, 'w') as f:
    f.write('''
def is_even(n):
    if n % 2 == 0:
        return True
    else:
        return False

def classify(n):
    if n < 0:
        return "negative"
    elif n == 0:
        return "zero"
    else:
        return "positive"
''')

json_out = os.path.join(tmp_dir, 'coverage.json')
cov = Coverage(source=[tmp_dir], branch=True, data_file=os.path.join(tmp_dir, '.coverage'))
cov.start()
sys.path.insert(0, tmp_dir)
sample = importlib.import_module('sample')
sample.is_even(4)
sample.classify(5)
cov.stop()
cov.save()
cov.json_report(outfile=json_out, include=[mod_path])

with open(json_out) as f:
    raw = json.load(f)

print(json.dumps(raw, indent=2))
```

```python
# Raw sys_info output
from coverage import Coverage
import json

cov = Coverage()
info = cov.sys_info()
print(json.dumps(dict(info), default=str, indent=2))
```

```python
# Run coverage on redditwarp via CLI and emit raw JSON report
import subprocess, os, json

rw = os.path.join(os.path.dirname(os.getcwd()), 'redditwarp')
cov_json = os.path.join(rw, 'coverage.json')

subprocess.run(
    ['python', '-m', 'coverage', 'run', '--branch', '--source=redditwarp',
     '-m', 'pytest', 'tests/', '--tb=no', '-q', '--co'],
    cwd=rw, timeout=120, capture_output=True
)
subprocess.run(
    ['python', '-m', 'coverage', 'json', '-o', cov_json],
    cwd=rw, capture_output=True
)

if os.path.exists(cov_json):
    with open(cov_json) as f:
        raw = json.load(f)
    print(json.dumps(raw, indent=2))
else:
    print('coverage.json not generated — run tests first')
```
