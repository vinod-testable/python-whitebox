# Bandit — Python Security Vulnerability Scanner
Install: `pip install bandit` | CLI: `bandit -r -f json <path>`

```python
import bandit
print(dir(bandit))
```

```python
import pkgutil, importlib
for finder, name, ispkg in pkgutil.walk_packages(bandit.__path__, bandit.__name__ + '.', onerror=lambda x: None):
    try:
        mod = importlib.import_module(name)
        print(name, '->', dir(mod))
    except BaseException as e:
        print(name, '-> SKIP:', e)
```

```python
# Raw JSON output from bandit on redditwarp
import subprocess, os

rw = os.path.join(os.path.dirname(os.getcwd()), 'redditwarp', 'redditwarp')
r = subprocess.run(
    ['bandit', '-r', '-f', 'json', rw],
    capture_output=True, text=True, encoding='utf-8', errors='replace', timeout=120
)
print(r.stdout)
```

```python
# Raw bandit output in SARIF format (structured)
import subprocess, os

rw = os.path.join(os.path.dirname(os.getcwd()), 'redditwarp', 'redditwarp')
r = subprocess.run(
    ['bandit', '-r', '-f', 'sarif', rw],
    capture_output=True, text=True, encoding='utf-8', errors='replace', timeout=120
)
print(r.stdout)
```
