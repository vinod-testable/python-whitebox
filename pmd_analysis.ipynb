# PMD — Static Code Analysis for Java

[PMD](https://pmd.github.io/) finds common programming flaws: unused variables, empty catch blocks, unnecessary object creation, etc.

## What PMD produces
- **Violations** with rule name, priority, file, line, description.
- Supports output formats: `text`, `xml`, `json`, `csv`, `sarif`, `html`.
- Ships with rule-sets in categories: `bestpractices`, `codestyle`, `design`, `documentation`, `errorprone`, `multithreading`, `performance`, `security`.

## CLI
```
pmd check -d <source-dir> -R <ruleset-csv> -f <format> [--report-file <file>]
```

## 1. Configuration

````python
from __future__ import annotations
import os, subprocess, sys
from pathlib import Path

_base = Path(r"F:\java_metrics")
PROJECT_ROOT = Path(os.environ.get("JAVA_PROJECT_ROOT", _base / "sample-java-app" / "src")).resolve()
PMD_HOME     = Path(os.environ.get("PMD_HOME", _base / "tools" / "pmd-bin-7.23.0")).resolve()
OUTPUT_DIR   = _base / "pmd_out"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

PMD_EXE = PMD_HOME / "bin" / ("pmd.bat" if sys.platform == "win32" else "pmd")

# Comma-separated rule-sets (uses PMD 7 category paths)
RULESETS = ",".join([
    "rulesets/java/quickstart.xml",
])

FORMATS = ["text", "xml", "json"]  # all three are generated

assert PMD_EXE.is_file(),      f"pmd not found: {PMD_EXE}"
assert PROJECT_ROOT.is_dir(),  f"PROJECT_ROOT not found: {PROJECT_ROOT}"
print(f"PMD_EXE      : {PMD_EXE}")
print(f"PROJECT_ROOT : {PROJECT_ROOT}")
print(f"OUTPUT_DIR   : {OUTPUT_DIR}")
print(f"RULESETS     : {RULESETS}")
````

## 2. List Java source files

````python
java_files = sorted(PROJECT_ROOT.rglob("*.java"))
print(f"Java files found: {len(java_files)}")
for f in java_files:
    print(" ", f.relative_to(PROJECT_ROOT))
````

## 3. Run PMD — raw output (text, xml, json)

````python
results = {}

for fmt in FORMATS:
    ext = {"text": "txt", "xml": "xml", "json": "json"}.get(fmt, fmt)
    report_file = OUTPUT_DIR / f"pmd-report.{ext}"
    cmd = [
        str(PMD_EXE), "check",
        "-d", str(PROJECT_ROOT),
        "-R", RULESETS,
        "-f", fmt,
        "--report-file", str(report_file),
    ]
    print(f"\n{'='*60}")
    print(f"Running PMD [{fmt}]: {' '.join(cmd)}")
    proc = subprocess.run(cmd, capture_output=True, text=True)
    print("STDOUT:", proc.stdout or "(empty)")
    print("STDERR:", proc.stderr or "(empty)")
    # PMD exits 4 when violations found, 0 when none, non-zero on error
    print("Exit code:", proc.returncode,
          "(4 = violations found, 0 = no violations)")
    results[fmt] = report_file

print("\nAll formats generated.")
````

## 4. Raw text report

````python
txt = results["text"]
if txt.is_file():
    print(f"=== {txt.name} ===")
    print(txt.read_text(encoding="utf-8", errors="replace") or "(empty — no violations)")
else:
    print("Text report not produced.")
````

## 5. Raw XML report

````python
xml = results["xml"]
if xml.is_file():
    print(f"=== {xml.name} ===")
    print(xml.read_text(encoding="utf-8", errors="replace"))
else:
    print("XML report not produced.")
````

## 6. Raw JSON report

````python
import json as _json
j = results["json"]
if j.is_file():
    raw = j.read_text(encoding="utf-8", errors="replace")
    print(f"=== {j.name} ===")
    try:
        print(_json.dumps(_json.loads(raw), indent=2))
    except Exception:
        print(raw)
else:
    print("JSON report not produced.")
````

## 7. All output files

````python
print("Files in output directory:")
for f in sorted(OUTPUT_DIR.iterdir()):
    print(f"  {f.name}  ({f.stat().st_size:,} bytes)")
````
